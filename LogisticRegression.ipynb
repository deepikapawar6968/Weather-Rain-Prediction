## spliting training and testing

```python
x = df.drop(['RainTomorrow'], axis=1)

y = df['RainTomorrow']
```

```python

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0) 
```

## handling null values

```python
cols1 = ["Evaporation", "Sunshine", "Cloud9am", "Cloud3pm"]

for col in cols1:

    # Location-wise median calculated ONLY from training data
    location_median = x_train.groupby("Location")[col].median()

    # Overall training median as fallback
    overall_median = x_train[col].median()

    # Fill training data
    x_train[col] = x_train[col].fillna(
        x_train["Location"].map(location_median)
    ).fillna(overall_median)

    # Fill test data using training medians
    x_test[col] = x_test[col].fillna(
        x_test["Location"].map(location_median)
    ).fillna(overall_median)

    print(f"Handled null values in {col}")
```

```python

cols2 = ["MinTemp","WindGustSpeed","Pressure9am","Pressure3pm","Temp9am","Temp3pm"]
for col in cols2:
    mean = x_train[col].mean()
    x_train[col]=x_train[col].fillna(mean)
    x_test[col] = x_test[col].fillna(mean)
    print(f"handled null values in {col}")
```

```python
cols3 = ["Rainfall","WindSpeed9am","WindSpeed3pm","Humidity9am","Humidity3pm","MaxTemp"]
for col in cols3:
    median = x_train[col].median()
    x_train[col]=x_train[col].fillna(median)
    x_test[col] = x_test[col].fillna(median)
    print(f"handled null values in {col}")
```

```python
cat_cols = ["WindGustDir", "WindDir9am", "WindDir3pm", "RainToday"]
for col in cat_cols:
    mode = x_train[col].mode()[0]
    x_train[col]=x_train[col].fillna(mode)
    x_test[col] = x_test[col].fillna(mode)
    print(f"handled null values in {col}")
```

```python
x_train.isnull().sum()
```

```python
x_test.isnull().sum()
```

## handling skewness

```python
log_cols = ["Rainfall", "Evaporation"]
```

```python
for col in log_cols:
    x_train[col] = np.log1p(x_train[col])
    x_test[col] = np.log1p(x_test[col])
```

```python
y_train = y_train.map({"No": 0, "Yes": 1})
y_test = y_test.map({"No": 0, "Yes": 1})
```

```python
x_train["RainToday"] = x_train["RainToday"].map({"No": 0, "Yes": 1})
x_test["RainToday"] = x_test["RainToday"].map({"No": 0, "Yes": 1})
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
cat_cols = ["Location", "WindGustDir", "WindDir9am", "WindDir3pm"]

encoder = OneHotEncoder(
    drop="first",
    handle_unknown="ignore",
    sparse_output=False
)

# Encode
x_train_cat = encoder.fit_transform(x_train[cat_cols])
x_test_cat = encoder.transform(x_test[cat_cols])

# Get column names
encoded_cols = encoder.get_feature_names_out(cat_cols)

# Convert encoded values into DataFrames
x_train_cat = pd.DataFrame(
    x_train_cat,
    columns=encoded_cols,
    index=x_train.index
)

x_test_cat = pd.DataFrame(
    x_test_cat,
    columns=encoded_cols,
    index=x_test.index
)

# Remove original categorical columns
x_train.drop(columns=cat_cols, inplace=True)
x_test.drop(columns=cat_cols, inplace=True)

# Add all encoded columns at once
x_train = pd.concat([x_train, x_train_cat], axis=1)
x_test = pd.concat([x_test, x_test_cat], axis=1)

print("Encoding completed")
```

## feature engeneering

```python
x_train["Date"] = pd.to_datetime(x_train["Date"])
x_test["Date"] = pd.to_datetime(x_test["Date"])

x_train["Year"] = x_train["Date"].dt.year
x_train["Month"] = x_train["Date"].dt.month
x_train["Day"] = x_train["Date"].dt.day

x_test["Year"] = x_test["Date"].dt.year
x_test["Month"] = x_test["Date"].dt.month
x_test["Day"] = x_test["Date"].dt.day
```

```python
x_train.drop(columns=["Date"], inplace=True)
x_test.drop(columns=["Date"], inplace=True)
```

## feature scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
print("Scaling completed")
```

## training model(logistic regression)

```python
# train a logistic regression model on the training set
from sklearn.linear_model import LogisticRegression


# instantiate the model
logreg = LogisticRegression(solver='liblinear', random_state=0)


# fit the model
logreg.fit(x_train, y_train)
```

## Compare model accuracy with null accuracy

- null accuracy : “How accurate would I be if I completely ignored all the features and simply predicted the majority class every time?”

## checking multiple thresholds

```python
from sklearn.preprocessing import binarize

#cm[row, column]

#Row = Actual , Column = Predicted

# TN = cm[0,0]
# TP = cm[1,1]
# FP = cm[0,1]
# FN = cm[1,0]
results = []

for threshold in [0.3, 0.4, 0.5, 0.6, 0.7]:

    y_pred = np.where(
        logreg.predict_proba(x_test)[:, 1] >= threshold,
        1,
        0
    )

    cm = confusion_matrix(y_test, y_pred)

    TN, FP, FN, TP = cm.ravel()

    recall = TP / (TP + FN)
    fpr = FP / (FP + TN)
    specificity = TN / (TN + FP)

    results.append({
        "Threshold": threshold,
        "Recall": round(recall, 4),
        "FPR": round(fpr, 4),
        "Specificity": round(specificity, 4)
    })

threshold_df = pd.DataFrame(results)

threshold_df
```
